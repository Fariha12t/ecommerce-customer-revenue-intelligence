```python
# Load order-item data from PostgreSQL

query = """
SELECT *
FROM fact_order_items;
"""

df = pd.read_sql(query, engine)

print("DATASET LOADED SUCCESSFULLY")
print("=" * 50)

print(f"Total rows: {df.shape[0]:,}")
print(f"Total columns: {df.shape[1]}")

display(df.head())
```

```python
# Inspect dataset structure

print("DATASET INFORMATION")
print("=" * 50)

df.info()

print("\nCOLUMN NAMES:")
print(df.columns.tolist())

print("\nSTATISTICAL SUMMARY:")
display(df.describe(include="all").T)
```

```python
# Check missing values and duplicates

print("DATA QUALITY REPORT")
print("=" * 50)

missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

display(missing.sort_values(
    "Missing Percentage",
    ascending=False
))

print("Duplicate rows:", df.duplicated().sum())

print("Unique orders:", df["order_id"].nunique())

print("Unique customers:", df["customer_unique_id"].nunique())

print("Unique products:", df["product_id"].nunique())
```

```python
# Initial business performance

delivered = df[
    df["order_status"] == "delivered"
].copy()

total_revenue = delivered["revenue_including_freight"].sum()

total_orders = delivered["order_id"].nunique()

total_customers = delivered["customer_unique_id"].nunique()

average_order_value = total_revenue / total_orders

print("ECOMMERCE BUSINESS OVERVIEW")
print("=" * 50)

print(f"Delivered Revenue: R$ {total_revenue:,.2f}")
print(f"Delivered Orders: {total_orders:,}")
print(f"Unique Customers: {total_customers:,}")
print(f"Average Order Value: R$ {average_order_value:,.2f}")
```

```python

# 4.1 Prepare sales data

# Keep delivered orders only
sales_df = df[
    df["order_status"].str.lower() == "delivered"
].copy()

# Convert timestamp to datetime
sales_df["order_purchase_timestamp"] = pd.to_datetime(
    sales_df["order_purchase_timestamp"],
    errors="coerce"
)

# Ensure revenue is numeric
sales_df["revenue_including_freight"] = pd.to_numeric(
    sales_df["revenue_including_freight"],
    errors="coerce"
)

# Remove records with invalid dates or revenue
sales_df = sales_df.dropna(
    subset=[
        "order_purchase_timestamp",
        "revenue_including_freight"
    ]
)

# Create monthly order-level revenue
order_level = (
    sales_df.groupby(
        [
            "order_id",
            pd.Grouper(
                key="order_purchase_timestamp",
                freq="MS"
            )
        ],
        as_index=False
    )
    .agg(
        order_revenue=("revenue_including_freight", "sum")
    )
)

print("Sales dataset prepared!")
print("Delivered order records:", len(sales_df))
print("Unique delivered orders:", order_level["order_id"].nunique())

display(order_level.head())
```

```python

# 4.2 Monthly revenue and order performance

monthly_sales = (
    order_level.groupby("order_purchase_timestamp")
    .agg(
        revenue=("order_revenue", "sum"),
        delivered_orders=("order_id", "nunique")
    )
    .reset_index()
)

monthly_sales["average_order_value"] = (
    monthly_sales["revenue"] /
    monthly_sales["delivered_orders"]
)

monthly_sales["revenue_growth_pct"] = (
    monthly_sales["revenue"].pct_change() * 100
)

monthly_sales = monthly_sales.round(2)

monthly_sales["month"] = (
    monthly_sales["order_purchase_timestamp"].dt.strftime("%Y-%m")
)

display(monthly_sales)
```

```python

# 4.3 Monthly revenue visualization

fig = px.line(
    monthly_sales,
    x="month",
    y="revenue",
    markers=True,
    title="Monthly Revenue Trend",
    labels={
        "month": "Month",
        "revenue": "Revenue (R$)"
    },
    template="plotly_white"
)

fig.update_layout(
    title_x=0.5,
    height=500,
    hovermode="x unified"
)

fig.show()
```

```python

# 5.1 Prepare customer-level order data

customer_orders = (
    sales_df.groupby(
        ["customer_unique_id", "order_id"],
        as_index=False
    )
    .agg(
        order_date=("order_purchase_timestamp", "max"),
        order_revenue=("revenue_including_freight", "sum")
    )
)

print("CUSTOMER ORDER DATA")
print("=" * 50)

print("Unique customers:", customer_orders["customer_unique_id"].nunique())
print("Delivered orders:", customer_orders["order_id"].nunique())

display(customer_orders.head())
```

```python

# 5.2 Calculate RFM metrics

reference_date = (
    customer_orders["order_date"].max()
    + pd.Timedelta(days=1)
)

customer_rfm = (
    customer_orders.groupby("customer_unique_id")
    .agg(
        recency_days=(
            "order_date",
            lambda x: (reference_date - x.max()).days
        ),
        frequency=("order_id", "nunique"),
        monetary=("order_revenue", "sum")
    )
    .reset_index()
)

customer_rfm["monetary"] = customer_rfm["monetary"].round(2)

print("RFM ANALYSIS")
print("=" * 50)

display(customer_rfm.head(10))
```

```python

# 5.3 Calculate RFM scores

# Recency: lower days = better score
customer_rfm["R_score"] = pd.qcut(
    customer_rfm["recency_days"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency: more orders = better score
customer_rfm["F_score"] = pd.qcut(
    customer_rfm["frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary: higher spending = better score
customer_rfm["M_score"] = pd.qcut(
    customer_rfm["monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

customer_rfm["RFM_score"] = (
    customer_rfm["R_score"]
    + customer_rfm["F_score"]
    + customer_rfm["M_score"]
)

display(customer_rfm.head(10))
```

```python

# 5.4 Segment customers

customer_rfm["customer_segment"] = np.select(
    [
        customer_rfm["RFM_score"] >= 13,
        customer_rfm["RFM_score"] >= 10,
        customer_rfm["RFM_score"] >= 7
    ],
    [
        "Champions",
        "Loyal Customers",
        "Potential Loyalists"
    ],
    default="At Risk / Others"
)

print("CUSTOMER SEGMENTATION")
print("=" * 50)

display(
    customer_rfm[
        [
            "customer_unique_id",
            "recency_days",
            "frequency",
            "monetary",
            "RFM_score",
            "customer_segment"
        ]
    ].head(15)
)
```

```python

# 5.5 Customer segment summary

segment_summary = (
    customer_rfm.groupby("customer_segment")
    .agg(
        customers=("customer_unique_id", "count"),
        total_revenue=("monetary", "sum"),
        average_spending=("monetary", "mean"),
        average_frequency=("frequency", "mean")
    )
    .reset_index()
)

segment_summary["customer_percentage"] = (
    segment_summary["customers"]
    / customer_rfm["customer_unique_id"].nunique()
    * 100
).round(2)

display(segment_summary.round(2))
```

```python

# 5.6 Customer segment distribution

fig = px.pie(
    segment_summary,
    names="customer_segment",
    values="customers",
    title="Customer Segment Distribution",
    hole=0.45,
    template="plotly_white"
)

fig.update_layout(title_x=0.5)

fig.show()
```

```python

# 5.7 Export customer analytics

customer_rfm.to_csv(
    output_dir / "customer_rfm_analysis.csv",
    index=False
)

segment_summary.to_csv(
    output_dir / "customer_segment_summary.csv",
    index=False
)

print("Customer analytics exported successfully!")

print("1. customer_rfm_analysis.csv")
print("2. customer_segment_summary.csv")
```
