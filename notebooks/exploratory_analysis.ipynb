# Exploratory Analysis — Olist E-Commerce

Run `python -m src.etl` and `python -m src.analytics` from the repository root first.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px

ROOT = Path.cwd()
REPORTS = ROOT / 'reports' / 'generated'
print('Working directory:', ROOT)
```

## Executive KPIs

```python
kpis = pd.read_csv(REPORTS / 'executive_kpis.csv')
kpis.T
```

## Monthly revenue trend

```python
monthly = pd.read_csv(REPORTS / 'monthly_performance.csv')
fig = px.line(monthly, x='month', y='revenue', markers=True,
              title='Monthly delivered-order gross value')
fig.update_layout(xaxis_title='Month', yaxis_title='Gross value (dataset currency)')
fig.show()
```

## Product category performance

```python
category = pd.read_csv(REPORTS / 'category_performance.csv').head(15)
px.bar(category.sort_values('revenue'), x='revenue', y='product_category_name_english',
       orientation='h', title='Top 15 categories by gross value').show()
```

## Customer segmentation

```python
rfm = pd.read_csv(REPORTS / 'customer_rfm.csv')
rfm.groupby('segment', dropna=False).agg(
    customers=('customer_unique_id', 'nunique'),
    average_monetary=('monetary', 'mean')
).sort_values('customers', ascending=False)
```

## Cohort retention

Values are proportions (0–1). Recent cohorts have fewer observable months.

```python
retention = pd.read_csv(REPORTS / 'cohort_retention.csv', index_col=0)
px.imshow(retention, aspect='auto', color_continuous_scale='Blues',
          title='Customer cohort retention').show()
```
